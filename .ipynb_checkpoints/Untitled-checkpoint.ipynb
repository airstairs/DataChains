```python
class nlist:

    def __init__(self, x=None):
        self.x = x
        self.next = None
        self.last = self

    def add(self, x):
        self.last.next = nlist(x)
        self.last = self.last.next
        
    def __repr__(self):
        if not self.next or self.next.x is None:
            return f"{self.x}"
        return f"{self.x} -> {self.next}"
    

import time
a = nlist()
timestart = time.time()
for i in range(10000000):
    a.add(i)
timeend = time.time()

timeend-timestart
```
