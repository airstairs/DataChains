```python
class myNumber:
    def __init__(self, number):
        self.number = number
        self.nextmyn = None
        
    def add(self, x):
        if self.nextmyn == None:
            self.nextmyn = myNumber(x)
            self.lastmyn = self.nextmyn
        else:
            self.lastmyn.nextmyn = myNumber(x)
            self.lastmyn = self.lastmyn.nextmyn
            
```

```python
x = myNumber(1)
```
