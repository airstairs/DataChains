```python
class nlist:

    def __init__(self, x=None):
        self.x = x
        self.next = None
        self.last = self

    def add(self, x):
        node = nlist(x)
        self.last.next = node
        self.last = node
    
    def __repr__(self):
        return (
                f"{self.x}"
                + (f" -> {self.next}" if self.next and self.next.x is not None else "")
            )
```

```python
a = nlist()
for i in range(20):
    a.add(i/2.3)
```
