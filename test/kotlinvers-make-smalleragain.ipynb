```kotlin
class nlist<T>(val x: T? = null) {
    var next: nlist<T>? = null
    var last: nlist<T> = this

    fun add(value: T) {
        val node = nlist(value)
        last.next = node
        last = node
    }

    override fun toString(): String {
        return "$x" + if (next != null) " -> $next" else ""
    }
}
```

```kotlin
val a = nlist<Any?>()
for (i in 0 until 20) {
    a.add(i / 2.3)
}
```

```kotlin
for (i in 1 until 10) {
    a.add("x".repeat(i))
}
```
