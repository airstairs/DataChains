```kotlin
class MyNumber<T>(var number: T) {
    var nextmyn: MyNumber<T>? = null
    var lastmyn: MyNumber<T>? = null

    fun add(x: T) {
        if (nextmyn == null) {
            nextmyn = MyNumber(x)
            lastmyn = nextmyn
        } else {
            lastmyn?.nextmyn = MyNumber(x)
            lastmyn = lastmyn?.nextmyn
        }
    }

    fun printAll() {
        var current: MyNumber<T>? = this
        while (current != null) {
            print("${current.number} ")
            current = current.nextmyn
        }
        println()
    }
}
```

```kotlin
val list = MyNumber(10)
list.add(20)
list.add(30)
list.add(40)

list.printAll() // Output: 10 20 30 40
```

```kotlin
val list1 =MyNumber(0*1.1)

for (i in 1..237) {
    list1.add(i*1.1)
}
list1.printAll()
```
